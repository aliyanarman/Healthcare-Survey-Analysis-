```python
import pandas as pd
```

```python
df = pd.read_csv("HCAHPS-National.csv")
df.head()
```

```python
df.shape 
```

```python
df.info()
```

```python
df.describe()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_") #modifying column names to improve data standards

df['start_date'] = pd.to_datetime(df['start_date'])
df['end_date'] = pd.to_datetime(df['end_date'])

df.isnull().sum()
```

```python
response_distribution = (                                #grouping responses in a separate dataframe for analysis
    df.groupby('hcahps_answer_description')['hcahps_answer_percent']
    .mean()
    .sort_values(ascending=False)
)

response_distribution             
```

```python
communication_df = df[      #creating a dataFrame that contains only questions related to nurse or doctor communication
    df['hcahps_question'].str.contains("nurse|doctor", case=False)
]

communication_df[['hcahps_question', 'hcahps_answer_description', 'hcahps_answer_percent']]
```

```python
top_metrics = df.sort_values('hcahps_answer_percent', ascending=False).head(10) # sorting postive responses in a descending order
bottom_metrics = df.sort_values('hcahps_answer_percent').head(10) #sorting negative responses in a descending order

top_metrics[['hcahps_question', 'hcahps_answer_percent']]

```

```python
bottom_metrics[['hcahps_question', 'hcahps_answer_percent']]

```

```python
strengths = (
    df[df['hcahps_answer_description'].str.contains("Always|Yes", case=False)]
    .sort_values('hcahps_answer_percent', ascending=False)
    .head(5)
)
                                             #sorting strengths and improvements needed
improvements = (
    df[df['hcahps_answer_description'].str.contains("Never|Sometimes|No", case=False)]
    .sort_values('hcahps_answer_percent', ascending=False)
    .head(5)
)

strengths[['hcahps_question', 'hcahps_answer_percent']] 
```

```python
improvements[['hcahps_question', 'hcahps_answer_percent']]
```
